```python
# Menghitung Rata-rata 3 Bilangan

A = int(input("Masukkan bilangan A: "))
B = int(input("Masukkan bilangan B: "))
C = int(input("Masukkan bilangan C: "))

RataRata = (A + B + C) / 3

print("Rata-rata =", RataRata)
```

```python
# Menghitung Nilai Fungsi

x = int(input("Masukkan nilai x (bukan 0): "))

if x != 0:
    f = 2 * x**3 + 2 * x + 15 / x
    print("Nilai f(x) =", f)
else:
    print("Error: x tidak boleh 0")
```

```python
# Menukar Nilai A, B, C, D Menjadi B, D, A, C

A = float(input("Masukkan nilai A: "))
B = float(input("Masukkan nilai B: "))
C = float(input("Masukkan nilai C: "))
D = float(input("Masukkan nilai D: "))

A, B, C, D = B, D, A, C

print("Nilai A =", A)
print("Nilai B =", B)
print("Nilai C =", C)
print("Nilai D =", D)
```

```python
# MEMBANDINGKAN DUA BILANGAN

x = int(input("Masukkan bilangan pertama X: "))
y = int(input("Masukkan bilangan kedua Y: "))

print("Bilangan pertama X =", x)
print("Bilangan kedua Y =", y)

if x > y:
    print("Bilangan terbesar adalah", x)
elif x < y:
    print("Bilangan terbesar adalah", y)
else:
    print("Kedua bilangan sama")
```

```python
# Menghitung Diskriminan Persamaan Kuadrat

a = 3
b = 5
c = 1

D = b**2 - 4*a*c

print("Diskriminan =", D)

if D < 0:
    print("Akar Imajiner")
elif D > 0:
    print("Akar Nyata Berlainan")
else:
    print("Akar Kembar")
```

```python
# Operator Logika

x = 5
n = 6

print(x > 0 and x < 10)
print(n % 2 == 0 or n % 3 == 0)
print(not (x > n))
```

```python
# Program Menentukan Kriteria Umur

umur = int(input("Masukkan umur Anda: "))

if umur <= 5:
    kriteria = "Balita"
elif umur <= 13:
    kriteria = "Anak-anak"
elif umur <= 25:
    kriteria = "Remaja"
elif umur <= 35:
    kriteria = "Dewasa"
elif umur <= 55:
    kriteria = "Orang Tua"
else:
    kriteria = "Lansia"

print("Umur Anda", umur, "tahun dan kriterianya yaitu", kriteria)
```

```python
# Program IF Bersarang

Nama = input("Nama Karyawan: ")
NIK = input("NIK Karyawan: ")
Status = input("Status (Staff/Honor): ")
Gol = input("Golongan (A/B/C): ")

if Status == "Staff":
    Gaji = 1000000
    if Gol == "A":
        Bonus = 200000
    elif Gol == "B":
        Bonus = 400000
    elif Gol == "C":
        Bonus = 550000
    else:
        Bonus = 0

elif Status == "Honor":
    Gaji = 750000
    if Gol == "A":
        Bonus = 150000
    elif Gol == "B":
        Bonus = 275000
    elif Gol == "C":
        Bonus = 480000
    else:
        Bonus = 0

else:
    Gaji = 0
    Bonus = 0

Total = Gaji + Bonus

print("Nama Karyawan:", Nama)
print("NIK Karyawan:", NIK)
print("Gaji = Rp", Gaji)
print("Bonus = Rp", Bonus)
print("Gaji Total = Rp", Total)
```
